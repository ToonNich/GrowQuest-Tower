```python
import os
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import classification_report, confusion_matrix

# Set path to test directory
test_dir = r'D:\Senior_WorkTable\GrowQuest_Data_Final\val'  # Make sure to use raw string for Windows paths

# Load the trained model
model = load_model('C:/Users/pondt/MainSenior/GrowQuest-Tower/Processing/hp_checkpoint/mobilenetv3_best_tuned_final.h5')

# Define parameters
img_height, img_width = 224, 224  # Set this to your model's input size

# Create data generator for test set
test_datagen = ImageDataGenerator(rescale=1./255)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(img_height, img_width),
    batch_size=32,
    class_mode='categorical',
    shuffle=False  # Important for correct predictions vs. filenames mapping
)

# Get metric names from the model
metrics = model.metrics_names
results = model.evaluate(test_generator)

print("\nEvaluation Results:")
for name, value in zip(metrics, results):
    print(f"{name}: {value:.4f}")


# Predict labels
predictions = model.predict(test_generator)
y_pred = predictions.argmax(axis=1)
y_true = test_generator.classes

# Classification report
print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=list(test_generator.class_indices.keys())))

# Optional: Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_true, y_pred))
```
